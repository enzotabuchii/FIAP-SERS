```python
# Criar um dataframe 'df4' a partir de dados somente com os registros de consumidores
# cuja demanda energética (Pot_Ativa) está acima de um limiar de 60% de demanda máxima, apenas
# com as colunas Data, Hora, Corrente e Pot_Ativa
df4 = dados[['Data', 'Hora', 'Corrente', 'Pot_Ativa']]
df4.head()
```

```python
# Qual é a demanda máxima (PMAX)?
PMAX = dados['Pot_Ativa'].max()
```

```python
print(f'A demanda máxima é: {PMAX:.2f} kW')
```

```python
# Qual é o limiar de 60% da demanda máxima (P60)?
P60 = 0.6 * PMAX
print(f'O limiar de 60% da demanda máxima é: {P60:.2f} kW')
```

```python
df4.columns
```

```python
df4 = df4[df4['Pot_Ativa'] > P60]
df4.head()
```

```python
print(f'Total de consumidores acima de P60: {df4.shape[0]}')
```
