# **Exemplo - Motor de Indução Trifásico**

### *1. Inserção dos dados de placa*

PU => Potência Útil (Mecânica)

n => Rendimento => n = PU / P

fp => Fator de Potência => fp = P / S

```python
PU = int(input('Insira o valor da Potência útil, em WATTS: '))
```

```python
n = float(input('Insira o rendimento, de 0 a 1: '))
```

```python
fp = float(input('Insira o fator de potência, de 0 a 1: '))
```

```python
PU, n, fp
```

### *2. Cálculo das potências Ativa (P), Aparente (S) e Reativa (Q)*

```python
# Potência Ativa (P) => n = PU / P
P = PU / n

# Potência Aparente (S) => fp = P / S
S = P / fp

# Potência Reativa (Q)
Q = (S**2 - P**2) ** 0.5
```

```python
P, S, Q
```

### *3. Apresentação dos resultados*

```python
print(f'A potência ativa será: {int(P)} W')
print(f'A potência Aparente será: {int(S)} VA')
print(f'A potência Reativa será: {int(Q)} VAr')
```
